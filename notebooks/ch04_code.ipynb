## 4.2　Python本体のバージョン

```bash
python3 --version        # 例: Python 3.10.12
```

## 4.3　仮想環境（venv）：標準の分け方

```bash
python3 -m venv .venv          # カレントに .venv という環境を作る
source .venv/bin/activate      # 有効化（Windowsは .venv\Scripts\activate）
# プロンプトの先頭に (.venv) が付けば成功。この環境の pip でパッケージをインストールする
pip install torch numpy pandas
deactivate                     # 環境から抜ける
```

## 4.4　conda / Miniconda：データサイエンスで広く使われる環境管理

```bash
conda create -n medai python=3.10     # medai という環境を Python3.10 で作る
conda activate medai                  # 有効化
conda install numpy pandas            # conda で入れる（pip との併用も可）
conda deactivate
```

## 4.5　pip・requirements.txt・uv：パッケージを入れる／固定する

```bash
pip install torch==2.4.0 monai nibabel      # バージョンを指定して入れる
pip freeze > requirements.txt               # 今の環境の全ライブラリを記録
pip install -r requirements.txt             # 別の環境で同じ構成を再現
```

## requirements.txt の実例と、バージョン固定の作法

```text
# requirements.txt ― 3次元CTセグメンテーションの例
torch==2.4.0
torchvision==0.19.0
monai==1.3.2          # 医療画像向けのディープラーニングのフレームワーク
nibabel==5.2.1        # NIfTI（Neuroimaging Informatics Technology Initiative、医療画像のファイル形式）(.nii.gz)の読み書き
pydicom==2.4.4        # DICOM（Digital Imaging and Communications in Medicine、医用画像の保存・通信の標準規格）の読み書き
numpy==1.26.4
scikit-image==0.24.0
matplotlib==3.9.2
```

## 4.7　GPU対応PyTorchのインストール：CUDAとの整合を確認する

```bash
pip install torch torchvision --index-url https://download.pytorch.org/whl/cu121
```

````python
import torch
print(torch.__version__)              # 例: 2.4.0+cu121  ← cuXXX がGPU対応の印
print(torch.cuda.is_available())      # True なら成功
````
